**4. VISUALISASIKAN NILAI UNIK**

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('/content/youtube top 100.csv')
unique_counts = df.nunique()
unique_counts.plot(kind="bar", color="blue", title=" Nilai Unik ")
plt.ylabel("Jumlah Nilai Unik")
plt.show()
```

**6. REPLACE NULL VALUES**

Karena pada dataset yang kami gunakan terdapat nilai kosong (null) pada kolom tags sebanyak 15 data, maka nilai kosong tersebut perlu dihapus.
Pada kasus ini, nilai kosong pada kolom tags kami ganti dengan string "No information"

```python
df['tags'] = df['tags'].fillna("No Information")  # Ganti NaN dengan angka 0
df
```

**10. CORRELATION**

   korelasi ini membantu kita memahami seberapa kuat dan dalam arah apa (positif atau negatif) variabel numerik saling berhubungan. Misalnya:
   - nilai korelasi yang mendekati 1 menunjukkan hubungan positif yang kuat
   - nilai yang mendekati -1 menunjukkan hubungan negatif yang kuat
   - nilai yang mendekati 0 menunjukkan hubungan yang lemah atau tidak ada hubungan linear.

```python
kolomNumerik = df[['view_count','duration','channel_follower_count']]
kolomNumerik.corr()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/youtube top 100.csv")
corr_matrix = df[['view_count', 'duration', 'channel_follower_count']].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Korelasi setiap kolom")
plt.show()
```
